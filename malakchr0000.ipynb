```python
import zipfile
import os

zip_path = '/content/Antibiotic Resistance Tracking Dataset.zip'
extract_path = '/content/antibiotic_resistance_data'

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Fichiers extraits :")
for root, dirs, files in os.walk(extract_path):
    for file in files:
        print(os.path.join(root, file))
```

```python
import pandas as pd

# Chemin vers le fichier CSV extrait
csv_path = '/content/antibiotic_resistance_data/Antibiotic Resistance Tracking Dataset/antibiotic_resistance_tracking.csv'

# Chargement des données
df = pd.read_csv(csv_path)

# Affichage des informations générales sur le dataset
print("--- Informations sur le Dataset ---")
df.info()

# Affichage d'un aperçu des premières lignes
print("\n--- Aperçu des 5 premières lignes ---")
display(df.head())

# Résumé statistique pour aider à identifier les variables X et Y
print("\n--- Nombre de valeurs uniques par colonne ---")
display(df.nunique())
```

```python
# Analyse et catégorisation des variables en X et Y

# Définition des rôles potentiels
variables_X = ['Age', 'Gender', 'Specimen_Type', 'Test_Method', 'Resistance_Genes']
variables_Y_drugs = ['Amoxicillin', 'Ciprofloxacin', 'Meropenem', 'Vancomycin', 'Colistin']
variables_Y_outcome = ['Outcome']

print("=== ANALYSE DE LA NATURE DES VARIABLES ===\n")

print("1. Variables Prédictrices Potentielles (X) :")
print("Ces variables représentent les caractéristiques du patient, de l'échantillon ou du test.\n")
for col in variables_X:
    unique_vals = df[col].dropna().unique()[:5]
    print(f"- {col} ({df[col].dtype}) : {len(df[col].dropna().unique())} valeurs uniques. Exemples : {list(unique_vals)}")

print("\n2. Variables Cibles Potentielles (Y) :\n")

print("A. Résistance aux antibiotiques (Classification de la résistance par médicament) :")
for col in variables_Y_drugs:
    unique_vals = df[col].dropna().unique()
    print(f"- {col} ({df[col].dtype}) : {list(unique_vals)}")

print("\nB. Issue clinique du patient (Prédiction de l'état de santé final) :")
for col in variables_Y_outcome:
    unique_vals = df[col].dropna().unique()
    print(f"- {col} ({df[col].dtype}) : {list(unique_vals)}")
```
